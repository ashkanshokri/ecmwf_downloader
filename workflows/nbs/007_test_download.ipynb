```python
from ecmwf_downloader.config.config import Config
from ecmwf_downloader.download import get_data
import sys
import os
from contextlib import contextmanager


@contextmanager
def suppress_stdout():
    with open(os.devnull, 'w') as devnull:
        old_stdout = sys.stdout
        sys.stdout = devnull
        try:
            yield
        finally:
            sys.stdout = old_stdout


with suppress_stdout():  # Suppress tqdm output
    config = Config()
    config['save_dir'] = '.'
    get_data(config)



```
