```python
from ecmwf_downloader.config.config import load_config
from ecmwf_downloader.download import get_data
from ecmwf_downloader.postprocess import get_save_dir
import xarray as xr
from tqdm import tqdm


config = load_config(f'X:\Projects\ECMWF\code\configs/msl.yaml')
config['save_dir'] = 'X:\Projects\ECMWF\cron_job_downloads'


bp = get_save_dir(config)
faulty_files = []

for t in config['type']:
    bpg = list(bp.glob(f'{t}*'))
    for p in tqdm(bpg):
        ds = xr.open_dataset(p)
        for param in config['param']:
            if ds[param].isnull().sum().values > 0 or 'time' not in ds[param].dims:
                faulty_files.append(p)
                print(len(faulty_files))
                break


print(f"{len(faulty_files)} out of {len(list(bp.glob(f'{t}*')))}  are faulty")
```

```python
import os
for f in faulty_files:
    try:    
        os.remove(f)
    except:
        pass
```
