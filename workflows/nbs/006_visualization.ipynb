```python
import json
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
from pathlib import Path
import pandas as pd
#config = load_config('X:\Projects\ECMWF\code\configs/tp.yaml')
save_dir = Path('X:/Projects/ECMWF/cron_job_downloads')

p = save_dir / 'tp' / 'downloaded_dates.json'

# Load JSON file
with open(p, "r") as file:
    date_strings = json.load(file)

# Convert to datetime objects and sort
dates = sorted(datetime.strptime(date, "%Y%m%d") for date in date_strings)

# Find missing dates
missing_dates = []
for i in range(len(dates) - 1):
    next_day = dates[i] + timedelta(days=1)
    while next_day < dates[i + 1]:
        missing_dates.append(next_day)
        next_day += timedelta(days=1)


df=pd.DataFrame(index=list(range(1,13)), columns=range(2024, 2026))
df[:]=0
for m, y in [(x.month, x.year) for x in missing_dates]:
    try:
        df.loc[m,y] += 1
    except:
        df.loc[m,y] = 1
df
```
